```python
# ==============================================================================
# AULA 1: INTRODUÇÃO A PYTHON & DESAFIO FINAL
# ==============================================================================
print("=" * 50)
print("AULA 1: INTRODUÇÃO A PYTHON")
print("=" * 50)

# 1. Primeiro teste de exibição
print("I competição de programação dá Start")

# 2. Resolução do Desafio: Criando variáveis de ano, local e tema
ano = "II"
local = "Start"
tema = "Programação"

# 3. Exibição da mensagem formatada usando f-string
print(f"{ano} competição de {tema} da {local}\n")


# ==============================================================================
# AULA 2: PONTUAÇÃO DE LEITURA & DESAFIO DA ANA E TOTAL
# ==============================================================================
print("=" * 50)
print("AULA 2: PONTUAÇÃO DE LEITURA")
print("=" * 50)

# 1. Definição da pontuação das categorias de livros
livro_ficcao = 8
livro_n_ficcao = 10
livro_infantil = 6

# Pontuação de Rodrigo (1 livro de cada categoria)
pontos_rodrigo = livro_ficcao + livro_n_ficcao + livro_infantil

# 2. Resolução do Desafio: Pontos de Ana (2 de ficção e 5 infantis)
pontos_ana = (2 * livro_ficcao) + (5 * livro_infantil)

# 3. Total de pontos acumulados por Rodrigo e Ana juntos
pontos_totais = pontos_rodrigo + pontos_ana

# 4. Exibição dos resultados
print(f"Os pontos totais do Rodrigo são: {pontos_rodrigo}")
print(f"Pontos acumulados por Ana: {pontos_ana}")
print(f"Total de pontos acumulados por Rodrigo e Ana juntos: {pontos_totais}\n")


# ==============================================================================
# AULA 3: DIVISÃO DE FIGURINHAS & DESAFIO COM EXTRAS
# ==============================================================================
print("=" * 50)
print("AULA 3: DIVISÃO DE FIGURINHAS")
print("=" * 50)

# 1. Coleta das entradas do usuário
total_figurinhas_iniciais = int(input("Digite o total de figurinhas no envelope: "))
figurinhas_extras = int(input("Digite a quantidade de figurinhas extras encontradas na mochila: "))
numero_de_amigos = int(input("Digite o número de amigos (sem contar o João): "))

# 2. Soma de todas as figurinhas
total_geral_figurinhas = total_figurinhas_iniciais + figurinhas_extras

# 3. Lógica de divisão das partes (Cada amigo = 1 parte; João = 2 partes)
figurinhas_por_amigo = total_geral_figurinhas // (numero_de_amigos + 2)
figurinhas_joao = 2 * figurinhas_por_amigo

# 4. Exibição dos resultados
print(f"Total geral de figurinhas: {total_geral_figurinhas}")
print(f"Cada amigo receberá: {figurinhas_por_amigo} figurinhas")
print(f"João recebeu: {figurinhas_joao} figurinhas\n")


# ==============================================================================
# AULA 4: CAPACIDADE DO BONDINHO & DESAFIO DOS 30 LUGARES COM PROFESSORES
# ==============================================================================
print("=" * 50)
print("AULA 4: CAPACIDADE DO BONDINHO")
print("=" * 50)

# 1. Coleta dos dados para o passeios
numero_de_alunos = int(input("Digite a quantidade de alunos: "))
numero_de_monitores = int(input("Digite a quantidade de monitores: "))
numero_de_professores = 8  # Quantidade fixa do desafio

# Mensagens de resposta
resposta_positiva = "pode ir"
resposta_negativa = "não pode ir"

# 2. Cálculo do total de passageiros
total_de_pessoas = numero_de_alunos + numero_de_monitores + numero_de_professores

# 3. Verificação de capacidade (Limite de 30 pessoas)
if total_de_pessoas <= 30:
    print(resposta_positiva)
else:
    print(resposta_negativa)
print("\n")


# ==============================================================================
# AULA 5: FLIPPER & DESAFIO DA PORTINHA Q
# ==============================================================================
print("=" * 50)
print("AULA 5: JOGO FLIPPER")
print("=" * 50)

# 1. Entrada de dados para as portas
P = int(input("Digite a posição da porta P (0 ou 1): "))
R = int(input("Digite a posição da porta R (0 ou 1): "))
Q = int(input("Digite a posição da porta Q (0 ou 1): "))

# 2. Estrutura condicional com prioridade para Q
if Q == 1:
    print("Caminho resultante: D")
elif P == 0:
    print("Caminho resultante: C")
elif R == 1:
    print("Caminho resultante: A")
else:
    print("Caminho resultante: B")
print("\n")


# ==============================================================================
# AULA 6: FUNÇÕES, IDADE DE DONA MÔNICA & DESAFIO DO LIVRO MAIS LONGO
# ==============================================================================
print("=" * 50)
print("AULA 6: FUNÇÕES E MAIOR VALOR")
print("=" * 50)

# 1. Função da Aula 6 (Dona Mônica)
def calcula_idade_maior(idade_monica, idade_filho1, idade_filho2):
    idade_filho3 = idade_monica - idade_filho1 - idade_filho2
    idade_maxima = max(idade_filho1, idade_filho2, idade_filho3)
    return idade_filho3, idade_maxima

# Execução de exemplo da aula
filho3, maior_idade = calcula_idade_maior(68, 12, 30)
print(f"Idade do 3º filho: {filho3} | Maior idade entre os três: {maior_idade}")

# 2. Resolução do Desafio Final da Aula 6 (Livro com mais páginas)
def maior_numero_de_paginas(p1, p2, p3, p4, p5):
    return max(p1, p2, p3, p4, p5)

livro_maior = maior_numero_de_paginas(120, 350, 210, 500, 180)
print(f"O livro com o maior número de páginas tem: {livro_maior} páginas\n")


# ==============================================================================
# AULA 7: TABUADA & DESAFIO DA SOMA DE 1 A 10
# ==============================================================================
print("=" * 50)
print("AULA 7: TABUADA E LAÇOS DE REPETIÇÃO")
print("=" * 50)

# 1. Função da aula (Tabuada de Multiplicação)
def calcula_tabuada(numero):
    print(f"--- Tabuada de Multiplicação do {numero} ---")
    for i in range(1, 11):
        print(f"{numero} x {i} = {numero * i}")

# 2. Resolução do Desafio Final da Aula 7 (Soma de 1 a 10)
def calcula_soma_1_a_10(numero):
    print(f"--- Tabuada de Soma do {numero} ---")
    for i in range(1, 11):
        print(f"{numero} + {i} = {numero + i}")

# Teste das funções da Aula 7
num_tabuada = int(input("Digite um número para ver as operações de 1 a 10: "))
calcula_tabuada(num_tabuada)
print()
calcula_soma_1_a_10(num_tabuada)
print("\n")


# ==============================================================================
# AULA 8: DESAFIO FINAL (CONTROLE DE ESTOQUE DE UMA LOJA)
# ==============================================================================
print("=" * 50)
print("AULA 8: DESAFIO FINAL - CONTROLE DE ESTOQUE")
print("=" * 50)

def controle_estoque():
    estoque = {"caderno": 10, "caneta": 20, "mochila": 5}
    print(f"Estoque inicial: {estoque}\n")

    produto = input("Digite o nome do produto (caderno, caneta, mochila): ").lower()
    operacao = input("Digite a operação ('entrada' ou 'saida'): ").lower()
    quantidade = int(input("Digite a quantidade: "))

    if produto in estoque:
        if operacao == "entrada":
            estoque[produto] += quantidade
            print(f"\nEntrada realizada com sucesso!")
            print(f"Estoque atualizado de {produto}: {estoque[produto]}")
        elif operacao == "saida":
            if estoque[produto] >= quantidade:
                estoque[produto] -= quantidade
                print(f"\nSaída realizada com sucesso!")
                print(f"Estoque atualizado de {produto}: {estoque[produto]}")
            else:
                print(f"\n[ERRO] Quantidade insuficiente! Estoque atual: {estoque[produto]}")
        else:
            print("\n[ERRO] Operação inválida! Digite 'entrada' ou 'saida'.")
    else:
        print("\n[ERRO] Produto não encontrado no estoque.")

# Execução do sistema de estoque
controle_estoque()
```
