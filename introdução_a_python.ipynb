```python
# ==============================================================================
# AULA 1: INTRODUÇÃO A PYTHON & DESAFIO FINAL
# ==============================================================================
print("=== AULA 1 ===")
# 1. Primeiro teste de exibição
print("I competição de programação dá Start")

# 2. Resolução do Desafio: Criando variáveis de ano, local e tema
ano = "II"
local = "Start"
tema = "Programação"

# 3. Exibição da mensagem formatada usando f-string
print(f"{ano} competição de {tema} da {local}\n")


# ==============================================================================
# AULA 2: PONTUAÇÃO DE LEITURA & DESAFIO DA ANA E TOTAL
# ==============================================================================
print("=== AULA 2 ===")
# 1. Definição da pontuação das categorias de livros
livro_ficcao = 8
livro_n_ficcao = 10
livro_infantil = 6

# Pontuação de Rodrigo (1 livro de cada categoria)
pontos_rodrigo = livro_ficcao + livro_n_ficcao + livro_infantil

# 2. Resolução do Desafio: Pontos de Ana (2 de ficção e 5 infantis)
pontos_ana = (2 * livro_ficcao) + (5 * livro_infantil)

# 3. Total de pontos acumulados por Rodrigo e Ana juntos
pontos_totais = pontos_rodrigo + pontos_ana

# 4. Exibição dos resultados
print(f"Os pontos totais do Rodrigo são {pontos_rodrigo}")
print(f"Pontos acumulados por Ana: {pontos_ana}")
print(f"Total de pontos acumulados por Rodrigo e Ana juntos: {pontos_totais}\n")


# ==============================================================================
# AULA 3: DIVISÃO DE FIGURINHAS & DESAFIO COM EXTRAS
# ==============================================================================
print("=== AULA 3 ===")
# 1. Coleta das entradas do usuário para a divisão de figurinhas
total_figurinhas_iniciais = int(input("Digite o total de figurinhas no envelope: "))
figurinhas_extras = int(input("Digite a quantidade de figurinhas extras encontradas na mochila: "))
numero_de_amigos = int(input("Digite o número de amigos (sem contar o João): "))

# 2. Soma de todas as figurinhas
total_geral_figurinhas = total_figurinhas_iniciais + figurinhas_extras

# 3. Lógica de divisão das partes (Cada amigo = 1 parte; João = 2 partes)
figurinhas_por_amigo = total_geral_figurinhas // (numero_de_amigos + 2)
figurinhas_joao = 2 * figurinhas_por_amigo

# 4. Exibição dos resultados das figurinhas
print(f"Total geral de figurinhas: {total_geral_figurinhas}")
print(f"Cada amigo receberá: {figurinhas_por_amigo} figurinhas")
print(f"João recebeu {figurinhas_joao} figurinhas\n")


# ==============================================================================
# AULA 4: CAPACIDADE DO BONDINHO & DESAFIO DOS 30 LUGARES COM PROFESSORES
# ==============================================================================
print("=== AULA 4 ===")
# 1. Coleta dos dados para o passeio do bondinho
numero_de_alunos = int(input("Digite a quantidade de alunos: "))
numero_de_monitores = int(input("Digite a quantidade de monitores: "))
numero_de_professores = 8  # Quantidade fixa de professores do desafio

# Mensagens de resposta
resposta_positiva = "pode ir"
resposta_negativa = "não pode ir"

# 2. Cálculo do total de passageiros
total_de_pessoas = numero_de_alunos + numero_de_monitores + numero_de_professores

# 3. Verificação de capacidade (Limite de 30 pessoas)
if total_de_pessoas <= 30:
    print(resposta_positiva)
else:
    print(resposta_negativa)
```
